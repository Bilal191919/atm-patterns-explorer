
# 🏧 ATM Transactions Data Analysis (Final Project Version)
This notebook performs **data cleaning, analysis, and visualization** on ATM transaction data using **Python, Pandas, and Matplotlib**.

---
## 📘 Step 1: Import Libraries and Load Dataset


```python

import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
try:
    file_path = 'atm_transactions.csv'
    atm_data = pd.read_csv(file_path)
    print("✅ Dataset loaded successfully.")
except FileNotFoundError:
    print("❌ Error: File not found. Please ensure 'atm_transactions.csv' is in the same folder.")
    raise
    
```


---
## 👀 Step 2: Preview the Dataset


```python

display(atm_data.head())
display(atm_data.tail())
    
```


---
## 🧹 Step 3: Data Cleaning and Translation
We’ll clean and translate column values for clarity.


```python

# Replace Turkish branch names and city names
atm_data['atmName'] = atm_data['atmName'].replace({
    'ALİAĞA BAĞLI ŞB': 'Allied Branch',
    'ALİAĞA İSTİKLAL CADDESİ': 'Allied Independence Street',
    'ALİAĞA PAZARYERİ': 'Allied Market',
    'ALİAĞA SAHİL PARKI': 'Allied Beach Park',
    'ALİAĞA ŞUBE-1': 'Allied Branch-1'
})

atm_data['atmCity'] = atm_data['atmCity'].replace({"Izmir": "Lahore"})

print("✅ Data cleaning completed successfully.")
    
```


---
## 📊 Step 4: View Cleaned Data


```python

display(atm_data[['atmId', 'atmName', 'atmCity', 'totalBalance',
                  'numberIncomeTransaction', 'numberOutcomeTransaction',
                  'totalIncome', 'totalOutcome', 'totalNumberTransaction',
                  'day', 'transactionTime']].head())
    
```


---
## 📈 Step 5: Total Transactions Analysis


```python

total_transactions = atm_data['totalNumberTransaction'].sum()
print(f"💰 Total Number of Transactions: {total_transactions}")

plt.figure(figsize=(8,4))
plt.bar(atm_data['atmName'], atm_data['totalNumberTransaction'], color='skyblue')
plt.title("Total Transactions per ATM")
plt.xlabel("ATM Name")
plt.ylabel("Total Number of Transactions")
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.show()
    
```


---
## 💵 Step 6: ATM Balance Analysis (Corrected Visualization)


```python

total_balance = atm_data['totalBalance'].sum()
print(f"🏦 Total Balance Across All ATMs: {total_balance}")

# Bar chart
plt.figure(figsize=(8,4))
plt.bar(atm_data['atmName'], atm_data['totalBalance'], color='orange')
plt.title("Total Balance per ATM")
plt.xlabel("ATM Name")
plt.ylabel("Total Balance (PKR)")
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.show()

# Histogram (Distribution)
plt.figure(figsize=(8,4))
plt.hist(atm_data['totalBalance'], bins=5, color='red', edgecolor='black')
plt.title("ATM Balance Distribution")
plt.xlabel("Balance Range (PKR)")
plt.ylabel("Number of ATMs")
plt.grid(alpha=0.3)
plt.show()
    
```


---
## 💾 Step 7: Export the Cleaned Dataset


```python

atm_data.to_csv('final_result.csv', index=False)
print("✅ Cleaned dataset exported as 'final_result.csv'")
    
```
